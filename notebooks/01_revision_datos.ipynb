```python

from pathlib import Path
import pandas as pd

# Directorio raíz del proyecto
BASE_DIR = Path.cwd()

#Comprobar si el script se está ejecutando desde la carpeta "notebooks"
if BASE_DIR.name == "notebooks":
    BASE_DIR = BASE_DIR.parent

DATA_DIR = BASE_DIR / "data"

print("Carpeta del proyecto:", BASE_DIR)
print("\nArchivos disponibles:")

for archivo in DATA_DIR.glob("*.csv"):
    print("-", archivo.name)
```

Carga de Datos

```python

# Cargar las dos fuentes existentes
peliculas = pd.read_csv(
    DATA_DIR / "netflix_movies_detailed_up_to_2025.csv"
)

series = pd.read_csv(
    DATA_DIR / "netflix_tv_shows_detailed_up_to_2025.csv"
)

# Revisar dimensiones
print("PELÍCULAS")
print("Filas y columnas:", peliculas.shape)
print("Columnas:", peliculas.columns.tolist())

print("\nSERIES")
print("Filas y columnas:", series.shape)
print("Columnas:", series.columns.tolist())
```

Exploración inicial 


```python

# 1. Resumen general de las dos fuentes

for nombre, df in [
    ("Películas", peliculas),
    ("Series", series)
]:
    print(f"\n{'=' * 50}")
    print(nombre.upper())
    print("=" * 50)

    print("Dimensiones:", df.shape)
    print("Registros duplicados:", df.duplicated().sum())
    print("IDs duplicados:", df["show_id"].duplicated().sum())
    print("Valores nulos:", df.isna().sum().sum())

    print("\nNulos por columna:")
    print(df.isna().sum().sort_values(ascending=False))

    print("\nTipos de datos:")
    print(df.dtypes)
```

Revisar la consistencia de los datos

```python

# 2. Revisar consistencia de las fuentes

for nombre, df in [
    ("Películas", peliculas),
    ("Series", series)
]:
    print(f"\n--- {nombre.upper()} ---")

    print("\nTipos de contenido:")
    print(df["type"].value_counts(dropna=False))

    print("\nRango de años:")
    print(df["release_year"].describe())

    print("\nValores financieros:")
    columnas_financieras = [
        c for c in ["budget", "revenue"] if c in df.columns
    ]

    if columnas_financieras:
        print(df[columnas_financieras].describe())

    print("\nEjemplos de registros:")
    display(df.head(3))
```

Verificar los duplicados y el origen de duration

```python

# A. Comprobar cuántos IDs están repetidos y cuántas filas involucran

conteo_ids = series["show_id"].value_counts()
ids_repetidos = conteo_ids[conteo_ids > 1]

print("Cantidad de IDs repetidos:", len(ids_repetidos))
print("Filas involucradas:", int(ids_repetidos.sum()))
print("\nFrecuencia de cada ID repetido:")
display(ids_repetidos.rename("cantidad_filas").to_frame())

# B. Comprobar si las filas son duplicados exactos
filas_duplicadas = series.duplicated(keep=False)

print("\nFilas que son duplicados exactos en todas las columnas:",
      int(series.duplicated().sum()))

print("\nFilas con show_id repetido:")
display(
    series[series["show_id"].isin(ids_repetidos.index)]
    .sort_values(["show_id", "title"])
)

# C. Investigar la columna duration de películas
print("\n=== DURACIÓN DE PELÍCULAS ===")
print("Tipo de dato:", peliculas["duration"].dtype)
print("Cantidad de valores distintos:",
      peliculas["duration"].nunique(dropna=True))
print("Primeros valores no nulos:")
print(peliculas["duration"].dropna().head(10).tolist())

# D. Comparar el estado de duration en ambas fuentes
print("\n=== DURACIÓN EN SERIES ===")
print("Tipo de dato:", series["duration"].dtype)
print("Valores nulos:", series["duration"].isna().sum())
print("Ejemplos de valores:")
print(series["duration"].dropna().head(10).tolist())
```

Diferencias dentro de cada grupo de identificadores repetidos

```python

# Auditoría detallada de los IDs repetidos en series

ids_repetidos = (
    series.loc[
        series["show_id"].duplicated(keep=False), "show_id"
    ]
    .unique()
)

for show_id in ids_repetidos:
    grupo = series[series["show_id"] == show_id]

    print(f"\n{'=' * 60}")
    print(f"ID: {show_id} | Filas: {len(grupo)}")

    columnas_diferentes = [
        columna
        for columna in series.columns
        if grupo[columna].nunique(dropna=False) > 1
    ]

    if columnas_diferentes:
        print("Columnas diferentes:", columnas_diferentes)
        display(grupo[["show_id"] + columnas_diferentes])
    else:
        print("Todas las columnas coinciden.")
```

Investigar por qué las películas no tienen duración

```python

# Auditoría de duración y campos relacionados en películas

columnas_revision = [
    "title", "release_year", "rating",
    "duration", "genres", "popularity",
    "vote_count", "vote_average", "budget", "revenue"
]

print("Valores nulos por columna:")
display(
    peliculas[columnas_revision]
    .isna()
    .sum()
    .sort_values(ascending=False)
    .to_frame("nulos")
)

print("\nEjemplos de películas:")
display(peliculas[columnas_revision].head(10))

print("\nRegistros con presupuesto o ingresos positivos:")
display(
    peliculas.loc[
        (peliculas["budget"] > 0) | (peliculas["revenue"] > 0),
        ["title", "duration", "budget", "revenue"]
    ].head(10)
)
```

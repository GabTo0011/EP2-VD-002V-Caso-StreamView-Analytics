```python

from pathlib import Path
import pandas as pd

# Directorio raíz del proyecto
BASE_DIR = Path.cwd()

#Comprobar si el script se está ejecutando desde la carpeta "notebooks"
if BASE_DIR.name == "notebooks":
    BASE_DIR = BASE_DIR.parent

DATA_DIR = BASE_DIR / "data"

print("Carpeta del proyecto:", BASE_DIR)
print("\nArchivos disponibles:")

for archivo in DATA_DIR.glob("*.csv"):
    print("-", archivo.name)
```

Carga de Datos

```python

# Cargar las dos fuentes existentes
peliculas = pd.read_csv(
    DATA_DIR / "netflix_movies_detailed_up_to_2025.csv"
)

series = pd.read_csv(
    DATA_DIR / "netflix_tv_shows_detailed_up_to_2025.csv"
)

# Revisar dimensiones
print("PELÍCULAS")
print("Filas y columnas:", peliculas.shape)
print("Columnas:", peliculas.columns.tolist())

print("\nSERIES")
print("Filas y columnas:", series.shape)
print("Columnas:", series.columns.tolist())
```

Exploración inicial 


```python

# 1. Resumen general de las dos fuentes

for nombre, df in [
    ("Películas", peliculas),
    ("Series", series)
]:
    print(f"\n{'=' * 50}")
    print(nombre.upper())
    print("=" * 50)

    print("Dimensiones:", df.shape)
    print("Registros duplicados:", df.duplicated().sum())
    print("IDs duplicados:", df["show_id"].duplicated().sum())
    print("Valores nulos:", df.isna().sum().sum())

    print("\nNulos por columna:")
    print(df.isna().sum().sort_values(ascending=False))

    print("\nTipos de datos:")
    print(df.dtypes)
```

Revisar ambos archivos y su calidad

```python

# 1. Resumen general de las dos fuentes

for nombre, df in [
    ("Películas", peliculas),
    ("Series", series)
]:
    print(f"\n{'=' * 50}")
    print(nombre.upper())
    print("=" * 50)

    print("Dimensiones:", df.shape)
    print("Registros duplicados:", df.duplicated().sum())
    print("IDs duplicados:", df["show_id"].duplicated().sum())
    print("Valores nulos:", df.isna().sum().sum())

    print("\nNulos por columna:")
    print(df.isna().sum().sort_values(ascending=False))

    print("\nTipos de datos:")
    print(df.dtypes)

```

Revisar la consistencia de los datos

```python

# 2. Revisar consistencia de las fuentes

for nombre, df in [
    ("Películas", peliculas),
    ("Series", series)
]:
    print(f"\n--- {nombre.upper()} ---")

    print("\nTipos de contenido:")
    print(df["type"].value_counts(dropna=False))

    print("\nRango de años:")
    print(df["release_year"].describe())

    print("\nValores financieros:")
    columnas_financieras = [
        c for c in ["budget", "revenue"] if c in df.columns
    ]

    if columnas_financieras:
        print(df[columnas_financieras].describe())

    print("\nEjemplos de registros:")
    display(df.head(3))
```

```python

print("=== PELÍCULAS ===")
print("Total de registros:", len(peliculas))
print("Duración nula:", peliculas["duration"].isna().sum())
print("Duración no nula:", peliculas["duration"].notna().sum())

print("\n=== SERIES ===")
print("Total de registros:", len(series))
print("IDs duplicados:", series["show_id"].duplicated().sum())

duplicados = series[
    series.duplicated("show_id", keep=False)
].sort_values("show_id")

display(
    duplicados[["show_id", "title", "release_year", "duration"]]
)

print("\n=== DATOS FINANCIEROS ===")
for columna in ["budget", "revenue"]:
    print(f"{columna}:")
    print("  Ceros:", (peliculas[columna] == 0).sum())
    print("  Nulos:", peliculas[columna].isna().sum())
    print("  Negativos:", (peliculas[columna] < 0).sum())
```
